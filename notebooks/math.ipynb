```python
def f1 (x:int)->int:
    return x*3-2

x=[1,4,5,7,2]
y=[]
for n in x:
    y.append(f1(n))

plt.plot(x,y)
plt.show()
```

```python
import math 

def f2(x:float)->float:
    # x^(pi/2)
    return math.pow(x, math.pi/2)


def f3(x: float) -> float:
    # x+sqrt(2x)
    return 5*x+math.sqrt(2*x)


xvals=[0,1,2,6,8,34]

yvals1 = []
yvals2 = []
for n in xvals:
    yvals1.append(f2(n))
    yvals2.append(f3(n))

plt.plot(xvals, yvals1)
plt.plot(xvals, yvals2)
plt.show()
```
