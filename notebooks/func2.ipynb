```python
import matplotlib.pyplot as plt

def ploteq(f,g,xmin=-10,xmax=10):
    x=np.linspace(xmin,xmax,400)
    y1=[f(n) for n in x]
    y2= [g(n) for n in x]

    plt.axhline(0,color="black")
    plt.axvline(0,color="black")

    plt.plot(x,y1,label="f(x)")
    plt.plot(x, y2, label="g(x)")

    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
from math import pi

f=lambda x:x**2
g=lambda x:x*pi
ploteq(f,g)
```

```python
f = lambda x: 2*x+3
g = lambda x: x+8
ploteq(f, g)
```
